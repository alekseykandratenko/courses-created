Los números deben ser introducidos por el usuario.

## 4. Listas


### Ejercicio 1: Gestión de lista de compras
Crea un programa que:
- Cree una lista vacía llamada "compras"
- Permita añadir 5 productos
- Muestre la lista
- Permita eliminar un producto
- Muestre la lista final

```python
compras = []

# Añadir 5 productos
for i in range(5):
    producto = input(f"Introduce el producto {i+1}: ")
    compras.append(producto)

print("\nLista de compras actual:", compras)
```

```python
# Eliminar un producto
producto_eliminar = input("¿Qué producto deseas eliminar? ")
if producto_eliminar in compras:
    compras.remove(producto_eliminar)
    print(f"Producto eliminado: {producto_eliminar}")
else:
    print("Producto no encontrado")

print("Lista final:", compras)
```

### Ejercicio 2: Operaciones con listas
Crea una lista de números y:
- Calcula la suma de todos los números
- Encuentra el número mayor y menor
- Ordena la lista de menor a mayor
- Elimina los números duplicados

```python
numeros = [4, 7, 2, 9, 1, 4, 3, 7, 8, 1]

# Suma de números
suma = sum(numeros)
print(f"Suma total: {suma}")

# Número mayor y menor
maximo = max(numeros)
minimo = min(numeros)
print(f"Número mayor: {maximo}")
print(f"Número menor: {minimo}")

# Ordenar lista
numeros_ordenados = sorted(numeros)
print(f"Lista ordenada: {numeros_ordenados}")

# Eliminar duplicados
numeros_sin_duplicados = list(set(numeros))
print(f"Lista sin duplicados: {numeros_sin_duplicados}")
```

### Ejercicio 1: Validador de edad
Crea un programa que:
- Pida la edad al usuario
- Determine si es:
  - Menor de edad (< 18)
  - Adulto (18-65)
  - Adulto mayor (> 65)

### Ejercicio 2: Calculadora de notas
Crea un programa que:
- Pida una calificación numérica (0-100)
- Muestre la calificación en letra:
   * A (90-100)
   * B (80-89)
   * C (70-79)
   * D (60-69)
   * F (0-59)

```python
calificacion = float(input("Introduce la calificación (0-100): "))

if calificacion >= 90:
    letra = "A"
elif calificacion >= 80:
    letra = "B"
elif calificacion >= 70:
    letra = "C"
elif calificacion >= 60:
    letra = "D"
else:
    letra = "F"

print(f"Tu calificación es: {letra}")
```

## 6. Bucles

### Ejercicio 1: Tabla de multiplicar
Crea un programa que:
- Pida un número al usuario
- Muestre su tabla de multiplicar del 1 al 10
- Use un bucle for

```python
numero = int(input("Introduce un número para ver su tabla de multiplicar: "))

for i in range(1, 11):
    resultado = numero * i
    print(f"{numero} x {i} = {resultado}")
```

### Ejercicio 2: Adivina el número
Crea un programa que:
- Genere un número aleatorio entre 1 y 100
- Pida al usuario que adivine el número
- De pistas (mayor o menor)
- Cuente los intentos
- Use un bucle while

```python
import random

numero_secreto = random.randint(1, 100)
intentos = 0

while True:
    intentos += 1
    numero_usuario = int(input("Adivina el número (1-100): "))
    
    if numero_usuario == numero_secreto:
        print(f"¡Correcto! Lo adivinaste en {intentos} intentos")
        break
    elif numero_usuario < numero_secreto:
        print("El número es mayor")
    else:
        print("El número es menor")
```
