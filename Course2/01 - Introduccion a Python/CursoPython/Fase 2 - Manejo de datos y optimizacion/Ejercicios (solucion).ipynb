### Ejercicio 2: Conjuntos
Dados dos conjuntos:
```python
A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7, 8}
```
* Encuentra la unión de A y B
* Encuentra la intersección de A y B
* Encuentra los elementos que están en A pero no en B

````python
A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7, 8}

# Unión
union = A | B  # También se puede usar A.union(B)
print(f"Unión: {union}")

# Intersección
interseccion = A & B  # También se puede usar A.intersection(B)
print(f"Intersección: {interseccion}")

# Diferencia (elementos en A pero no en B)
diferencia = A - B  # También se puede usar A.difference(B)
print(f"Diferencia (A-B): {diferencia}")
````

````python
biblioteca = {
    "Ficción": {
        "1984": {
            "autor": "George Orwell",
            "año": 1949,
            "disponible": True
        },
        "El Quijote": {
            "autor": "Miguel de Cervantes",
            "año": 1605,
            "disponible": False
        }
    },
    "No ficción": {
        "Sapiens": {
            "autor": "Yuval Noah Harari",
            "año": 2011,
            "disponible": True
        }
    },
    "Ciencia": {
        "Breve historia del tiempo": {
            "autor": "Stephen Hawking",
            "año": 1988,
            "disponible": True
        }
    }
}

# Ejemplo de uso
def buscar_libro(titulo):
    for categoria, libros in biblioteca.items():
        if titulo in libros:
            info = libros[titulo]
            estado = "disponible" if info["disponible"] else "no disponible"
            return f"{titulo} escrito por {info['autor']} ({info['año']}) - {estado}"
    return "Libro no encontrado"

print(buscar_libro("1984"))
print(buscar_libro("Sapiens"))
````

### Ejercicio 4: Pilas y Colas
Implementa:
* Una pila (LIFO) usando una lista con métodos push() y pop()
* Una cola (FIFO) usando una lista con métodos enqueue() y dequeue()

````python
class Pila:
    def __init__(self):
        self.items = []
    
    def push(self, item):
        self.items.append(item)
    
    def pop(self):
        if not self.esta_vacia():
            return self.items.pop()
        return None
    
    def esta_vacia(self):
        return len(self.items) == 0
    
    def __str__(self):
        return str(self.items)

class Cola:
    def __init__(self):
        self.items = []
    
    def enqueue(self, item):
        self.items.append(item)
    
    def dequeue(self):
        if not self.esta_vacia():
            return self.items.pop(0)
        return None
    
    def esta_vacia(self):
        return len(self.items) == 0
    
    def __str__(self):
        return str(self.items)

# Ejemplo de uso
print("=== Prueba de Pila ===")
pila = Pila()
pila.push(1)
pila.push(2)
pila.push(3)
print(f"Pila: {pila}")
print(f"Pop: {pila.pop()}")
print(f"Pila final: {pila}")

print("\n=== Prueba de Cola ===")
cola = Cola()
cola.enqueue(1)
cola.enqueue(2)
cola.enqueue(3)
print(f"Cola: {cola}")
print(f"Dequeue: {cola.dequeue()}")
print(f"Cola final: {cola}")
````

### Ejercicio 1: Gestión de archivos
Crea un programa que:
* Solicite al usuario nombres y edades de estudiantes
* Guarde esta información en un archivo 'estudiantes.txt'
* Lea el archivo y muestre los estudiantes ordenados por edad

````python
print("Programa para guardar datos de estudiantes")
print("Para terminar, escribe 'fin' como nombre")
print("----------------------------------------")

# Lista para guardar los estudiantes
estudiantes = []

# Pedimos datos hasta que escriban 'fin'
while True:
    nombre = input("\nEscribe el nombre (o 'fin' para terminar): ")
    
    # Si escribe 'fin', salimos del bucle
    if nombre.lower() == "fin":
        break
    
    # Pedimos y validamos la edad (asegurándonos que sea un número)
    while True:
        edad = input(f"Escribe la edad de {nombre}: ")
        if edad.isdigit():  # Comprobamos que sean solo números
            break
        print("Por favor, escribe un número válido para la edad")
    
    # Añadimos el estudiante a la lista
    estudiantes.append([nombre, edad])
    print("¡Estudiante guardado!")

# Si hay estudiantes, los guardamos en el archivo
if estudiantes:
    # Guardamos en el archivo
    with open("estudiantes.txt", "w") as archivo:
        for estudiante in estudiantes:
            archivo.write(f"{estudiante[0]},{estudiante[1]}\n")
    
    print("\nEstudiantes guardados:")
    print("----------------------------------------")
    
    # Mostramos los estudiantes guardados
    for estudiante in estudiantes:
        print(f"Nombre: {estudiante[0]}, Edad: {estudiante[1]} años")
else:
    print("\nNo se guardaron estudiantes")
````

### Ejercicio 1: Funciones recursivas
Implementa las siguientes funciones recursivas:
* Factorial de un número
* Fibonacci hasta n términos
* Suma de dígitos de un número

````python
def factorial(n):
    if n < 0:
        raise ValueError("El factorial no está definido para números negativos")
    if n <= 1:
        return 1
    return n * factorial(n - 1)

def fibonacci(n):
    if n <= 0:
        return []
    elif n == 1:
        return [0]
    elif n == 2:
        return [0, 1]
    
    fib = fibonacci(n - 1)
    fib.append(fib[-1] + fib[-2])
    return fib

def suma_digitos(n):
    if n < 10:
        return n
    return n % 10 + suma_digitos(n // 10)

# Pruebas
try:
    # Factorial
    print("=== Factorial ===")
    for i in range(6):
        print(f"{i}! = {factorial(i)}")
    
    # Fibonacci
    print("\n=== Fibonacci ===")
    n_terms = 10
    print(f"Primeros {n_terms} términos: {fibonacci(n_terms)}")
    
    # Suma de dígitos
    print("\n=== Suma de dígitos ===")
    numeros = [123, 45678, 9]
    for num in numeros:
        print(f"Suma de dígitos de {num} = {suma_digitos(num)}")
        
except ValueError as e:
    print(f"Error: {e}")
````

### Ejercicio 2: Argumentos variables
Crea una función calculadora que:
* Acepte un número variable de argumentos
* Permita especificar la operación como parámetro nombrado
* Maneje las operaciones: suma, resta, multiplicación, división

````python
# Solución Ejercicio 2
def calculadora(*args, operacion='suma'):
    if len(args) == 0:
        raise ValueError("Se requiere al menos un número")
    
    operaciones = {
        'suma': lambda x, y: x + y,
        'resta': lambda x, y: x - y,
        'multiplicacion': lambda x, y: x * y,
        'division': lambda x, y: x / y if y != 0 else float('inf')
    }
    
    if operacion not in operaciones:
        raise ValueError(f"Operación no válida. Opciones: {', '.join(operaciones.keys())}")
    
    resultado = args[0]
    for num in args[1:]:
        resultado = operaciones[operacion](resultado, num)
    
    return resultado

# Pruebas
try:
    # Suma
    print(f"Suma: {calculadora(1, 2, 3, 4, 5)}")
    
    # Resta
    print(f"Resta: {calculadora(100, 20, 30, operacion='resta')}")
    
    # Multiplicación
    print(f"Multiplicación: {calculadora(2, 3, 4, operacion='multiplicacion')}")
    
    # División
    print(f"División: {calculadora(100, 2, 2, operacion='division')}")
    
    # Prueba de error
    print(calculadora(1, 2, operacion='potencia'))
    
except ValueError as e:
    print(f"Error: {e}")
except ZeroDivisionError:
    print("Error: División por cero")
````

## 4. Gestión de errores

### Ejercicio 1: Manejo de excepciones
Crea una función división_segura que:
* Acepte dos números como parámetros
* Maneje las excepciones: ZeroDivisionError, TypeError, ValueError
* Devuelva el resultado de la división o un mensaje de error apropiado

````python
def division_segura(numerador, denominador):
    try:
        # Convertir a float en caso de que sean strings
        num = float(numerador)
        den = float(denominador)
        
        # Intentar división
        resultado = num / den
        return f"Resultado: {resultado:.2f}"
    
    except ZeroDivisionError:
        return "Error: No es posible dividir entre cero"
    
    except ValueError:
        return "Error: Por favor, introduce números válidos"
    
    except TypeError:
        return "Error: Tipo de datos no válido"
    
    except Exception as e:
        return f"Error inesperado: {str(e)}"

# Pruebas
pruebas = [
    (10, 2),        # División normal
    (10, 0),        # División por cero
    ('10', '2'),    # Strings numéricos válidos
    ('diez', 2),    # String no numérico
    (10, None),     # Tipo no válido
]

for n, d in pruebas:
    print(f"\nDividiendo {n} entre {d}:")
    print(division_segura(n, d))
````

### Ejercicio 2: Excepciones personalizadas
Crea una clase de excepción personalizada llamada EdadInvalidaError y úsala en una función que valide la edad de una persona para un club nocturno (18-99 años).

````python
# Solución Ejercicio 10
class EdadInvalidaError(Exception):
    """Excepción que se lanza cuando la edad está fuera del rango permitido."""
    pass

def validar_edad_club(edad):
    try:
        edad = int(edad)
        
        if edad < 18:
            raise EdadInvalidaError("Debes ser mayor de edad para entrar al club")
        elif edad > 99:
            raise EdadInvalidaError("La edad ingresada excede el límite permitido")
        
        return f"Bienvenido al club, tienes {edad} años"
    
    except ValueError:
        return "Error: Por favor, introduce un número válido"
    
    except EdadInvalidaError as e:
        return f"Acceso denegado: {str(e)}"

# Pruebas
edades_prueba = [17, 25, 100, "veinte", 45]

for edad in edades_prueba:
    print(f"\nValidando edad: {edad}")
    print(validar_edad_club(edad))
````
