```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_ollama.llms import OllamaLLM
from langchain_core.output_parsers import StrOutputParser
```

# Implementación básica

```python
# define model name to interact with
model_name = "llama3.1"

llm = OllamaLLM(
    model=model_name,   # model name
    temperature=0,      # model temperature (increase value to make answer be more abstract)
    # format="json",    # output format
    base_url='http://127.0.0.1:11434',   # model serve url
    keep_alive=True     # keep connection alive with llm model server
)
```

```python
print(llm.invoke('Dime que sabes sobre la luna.'))
```

# Ejemplo de definición de prompt especifico

```python
def medical_asystant(question):
    template = """Actua como experto médico de medecina general y crea una respuesta detallata en función de la pregunta.
        Pregunta: {question}"""
    
    prompt = ChatPromptTemplate.from_template(template)

    chain = prompt | llm | StrOutputParser()

    result = chain.invoke({"question": question})

    return result
```

```python
print(medical_asystant('Tengo un dolor de cabeza y temperatura corporal de 38 grados. ¿Qué me pasa y que debo hacer?'))
```

# Implementación de chat con historial

```python
def create_prompt(question, chat_history):
    template = """Actua como experto médico de medicina general y crea una respuesta detallada en función de la pregunta.
    Historial:
    {chat_history}
    Pregunta: {question}"""

    return template.format(chat_history=chat_history, question=question)

def interact_with_model(question, chat_history):
    prompt = create_prompt(question, chat_history)
    prompt_template = ChatPromptTemplate.from_template(prompt)
    chain = prompt_template | llm | StrOutputParser()

    resultado = chain.invoke({"question": question})
    
    return resultado
```

```python
chat_history = []
```

```python
while True:
    pregunta = input("Escribe tu pregunta: ")
    print(f"Usuario: {pregunta}\n")
    if pregunta.lower() == "salir":
        break
    respuesta = interact_with_model(pregunta, "\n".join(chat_history))
    chat_history.append(f"Usuario: {pregunta}")
    chat_history.append(f"Modelo: {respuesta}")
    print(f"Modelo: {respuesta}\n")
```
